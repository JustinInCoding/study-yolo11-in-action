```python
from ultralytics import YOLO
model = YOLO('./weights/yolo11n.pt')
# 保存带标注的可视化图片
# 保存检测结果的文本文件, txt, YOLO 格式 - 类别 x_center y_center width height （归一化后的数值）
results = model.predict(source = './images/bee.jpg', save = True, save_txt = True)
```
