```python
# 导入 HTML 类以显示 HTML 内容
from IPython.display import HTML
# 导入 b64encode 用于将二进制数据编码为 base64
from base64 import b64encode

# 读取 mp4 文件内容
mp4 = open('runs/detect/track/video.mp4', 'rb').read()
# 将 mp4 文件内容编码为 base64，并拼接成数据 url
data_url = "data:video/mp4;base64," + b64encode(mp4).decode()

# 再次导入 HTML 类和 display 函数，用于显示 HTML 内容和视频
from IPython.display import HTML, display

# 创建 HTML 字符串，嵌入视频源
html_video = HTML(
"""
<video controls>
    <source src = '%s' type = "video/mp4" />
    Your browser does not support the video tag.
</video>
""" % data_url
)

# 显示视频
display(html_video)
```
